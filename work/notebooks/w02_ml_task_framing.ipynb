# 2. Target or Proxy

The starter dataset does not directly provide a "refresh priority" column.

Pages with declining search performance represent stronger candidates for content refresh, while stable or growing pages have lower refresh priority.

The long-term goal is to build a model that assigns every page a refresh opportunity score.

# 3. Success Metric

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

# 5. Why ML Beats a Fixed Rule

A simple rule such as "refresh every page with CTR below 1%" ignores many other important signals.

Machine learning can consider multiple features simultaneously, including click-through rate, average position, engagement rate, search volume, and content characteristics.

By learning patterns across many pages, a machine learning model can produce more useful rankings than a single fixed threshold.
