## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

## Research Question

### Decision
Which content pages should be reviewed first for refresh, expansion, protection, pruning, or monitoring?

### Who acts?
The SEO manager or content editor.

### Action
The SEO manager or content editor reviews the highest-priority pages and decides whether they should be refreshed, expanded, monitored, protected, or pruned.

### Cost of a wrong recommendation
If low-priority pages are recommended first, editors may waste valuable time while important pages continue losing visibility and traffic. Missing high-priority pages can also reduce future business opportunities.

### Why machine learning?
A simple rule based on one metric, such as CTR or impressions, is often not enough. Multiple search and engagement signals interact, so machine learning can help rank pages more effectively for review.
